```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Pengaturan Jumlah Data
jumlah_baris = 150000
wilayah = ["Makassar", "Gowa", "Maros", "Takalar", "Pangkep", "Jeneponto"]
bobot_wilayah = [0.55, 0.20, 0.10, 0.05, 0.05, 0.05]
layanan = ["Sedot WC", "Pelancaran Pipa Mampet"]
status = ["Selesai", "Batal", "Reschedule"]
bobot_status = [0.85, 0.05, 0.10]

data = []
tanggal_mulai = datetime(2025, 1, 1)

print("Mulai membuat 150.000 data transaksi Jasa Sedot WC MGM...")

for i in range(jumlah_baris):
    id_pesanan = f"MGM-25{str(i).zfill(5)}"
    hari_acak = random.randint(0, 365)
    jam_acak = random.randint(7, 17)
    menit_acak = random.randint(0, 59)
    waktu_order = tanggal_mulai + timedelta(days=hari_acak, hours=jam_acak, minutes=menit_acak)

    w_pilih = random.choices(wilayah, weights=bobot_wilayah)[0]
    l_pilih = random.choice(layanan)
    s_pilih = random.choices(status, weights=bobot_status)[0]

    if l_pilih == "Sedot WC":
        harga = random.choice([350000, 400000, 450000])
    else:
        harga = random.choice([200000, 250000, 300000])

    if s_pilih != "Selesai":
        harga = 0

    rating = random.choices([3, 4, 5], weights=[0.1, 0.3, 0.6])[0] if s_pilih == "Selesai" else 0

    data.append([id_pesanan, waktu_order.strftime("%Y-%m-%d %H:%M:%S"), l_pilih, w_pilih, s_pilih, harga, rating])

df = pd.DataFrame(data, columns=["id_pesanan", "tanggal_waktu", "jenis_layanan", "wilayah_operasional", "status_pekerjaan", "nilai_transaksi", "rating_pelanggan"])
df.to_csv("dataset_sedot_wc_mgm.csv", index=False)
print("BERHASIL! File dataset_sedot_wc_mgm.csv sudah dibuat.")
```
